# How much amount the companies spent on R & D ?

```python
df.head()
```

```python
df.columns
```

```python
print('Comapny spending for R & D in $Bn')
RD = df.groupby('Company')['R&D_Spending_USD_Mn'].sum()/100
RD
```

```python
# Draw a Bar Plot to Show the amount Spent R & D by the companies

plt.bar(RD.index, RD.values, color = ['cyan', 'black', 'magenta'])

plt.title('R&D Spending by the Companies')
plt.xlabel('Comapny')
plt.ylabel('Amount in USD_$Bn')

plt.show()
```

# Revenue Earned by the Companies

```python
df.head()
```

```python
rev = df.groupby('Company')['AI_Revenue_USD_Mn'].sum()/1000

rev
```

```python
# Draw a Bar Plot to Show the Revenue of the companies

plt.bar(rev.index, rev.values, color = ['cyan', 'black', 'magenta'], width= 0.2)

plt.title('Revenue earned by the companies')
plt.xlabel('Comapny')
plt.ylabel('Amount in USD_$Bn')

plt.show()
```

```python
# Barplots to show expenditure & revenue of the companies

plt.figure(figsize=(10,4))

plt.subplot(1,2,1)

plt.bar(RD.index, RD.values, color = ['cyan', 'black', 'magenta'])

plt.title('R&D Spending by the Companies')
plt.xlabel('Comapny')
plt.ylabel('Amount in USD_$Bn')


plt.subplot(1,2,2)

plt.bar(rev.index, rev.values, color = ['cyan', 'black', 'magenta'], width= 0.2)

plt.title('Revenue earned by the companies')
plt.xlabel('Comapny')
plt.ylabel('Amount in USD_$Bn')

plt.show()
```

Expenditure vs Revenue year-by-year

```python
df.head()
```

```python
spend = df.groupby('year')['R&D_Spending_USD_Mn'].sum()
spend
```

```python
# Showing the Amount spent on R & D
plt.plot(spend.index, spend.values, color = 'r')

plt.title('Combined R&D Spending year-by-year')
plt.xlabel('Year')
plt.ylabel('Amount in USD-$Bn')

plt.show()
```

```python
revenue = df.groupby('year')['AI_Revenue_USD_Mn'].sum()
revenue
```

```python
# Showing the Revenue earnd
plt.plot(revenue.index, revenue.values, color = 'g')

plt.title('Combined R&D Spending year-by-year')
plt.xlabel('Year')
plt.ylabel('Amount in USD-$Bn')

plt.show()
```

```python
plt.plot(spend.index, spend.values, color = 'r')
plt.plot(revenue.index, revenue.values, color = 'g')

plt.title('Combined Expenditure vs Revenue year-by-year', fontsize= 10)
plt.xlabel('Year')
plt.ylabel('Amount in USD-$Bn')

plt.legend(['Expenditure', 'Revenue'])

plt.show()
```
